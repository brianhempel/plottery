## Data (do not change)

```python
groups = ['control', 'experiment']

durations = [
    [18, 15.42, 25.55, 12.56, 8.67],
    [7.31, 9.66, 13.64, 14.92, 18.47]
] # durations[0] corresponds to the `control` group, durations[1] corresponds to the `experiment` group

success = [
    [1, 1, 0, 0, 1],
    [0, 0, 1, 1, 1]
] # success[0] corresponds to the `control` group, success[1] corresponds to the `experiment` group

time_limit = 30
```

Imagine the data above to come from a controlled experiment with subjects:
To understand the data above, let's focus on `durations[0][0]` (18) and `success[0][0]` (1).
- This means that this subject is in the control group, finished the task in 18 minutes, and the attempt succeeded.
If we look at `durations[1][1]` (9.66) and `success[1][1]` (0).
- This means that this subject is in the experiment group, finished the task in 9.66 minutes, but the attempt failed.
