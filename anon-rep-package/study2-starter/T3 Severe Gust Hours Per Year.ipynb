## Severe Gust Hours Per Year

The goal is to plot the average number of hours in a year that a location experiences a severe wind gust (defined as ≥50kts for 3 seconds). The data comes from ASOS weather stations scattered throughout the United States.

The dots represent the gustiness at individual weather stations. The background shading is an interpolation of nearby stations.

<img src="../study_2_goals/ASOS_Gust_Hours_per_Year_2003-2021.png" width="640">

**Notes:**
- You do not need to use any special mapping library or projetions, plain latitude/longitude plotting and `imshow` is fine.
- The map consists of three layers in this order:
    - A "blurred" interpolation of the data across nearby weather stations to provide a contoured background.
    - The political boundaries.
    - The scatterplot dots representing the individual weather stations, colored by their gustiness.
- You do not need to do any fancy calculation, the above are provided.
- You do not have to use the colors above.
- But, pay close attention to the color mapping.  A discrete color scheme (instead of continuous) can help. Color maps are a little wonky in matplotlib and it's easy to get the colors desynced from the data, in part because matplotlib divides color maps into the colormap and a "norm" that maps numbers to the color scale.
