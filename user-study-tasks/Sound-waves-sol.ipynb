## Data (do not change)

```python
# Parameters
sampling_rate = 44100  # Samples per second
frequency = 440  # Frequency of the sine wave in Hz
duration = 1  # Duration in seconds

# Time array
t = np.linspace(0, duration, int(sampling_rate * duration), endpoint=False)[:200]

### Sine wave
sine_wave = np.sin(2 * np.pi * frequency * t).tolist()

### Square wave
square_wave = np.sign(np.sin(2 * np.pi * frequency * t) + 0.00000001).tolist()

### Triangle wave
triangle_wave = 2 * np.abs(2 * (t * frequency - np.floor(t * frequency + 0.5))) - 1

### Sawtooth wave
sawtooth_wave = 2 * (t * frequency - np.floor(t * frequency + 0.5))
```
