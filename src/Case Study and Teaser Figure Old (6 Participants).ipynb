```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
participants = ('P1', 'P2', 'P3', 'P4', 'P5', 'P6', 'Mean')

interaction_counts = { 'Code': (3,  6, 12, 1,  14, 4,  6.7),
                       'GUI':  (12, 0, 4,  2,  4,  17, 6.5),
                       'AI':   (5,  7, 4,  11, 4,  4,  5.8)  }

interaction_percentages = { 'Code': (6.2,  27.6, 55.3, 0.4,  27.1, 19.6, 22.7),
                            'GUI':  (46.7, 0.0,  18.0, 6.3,  11.0, 35.6, 19.6),
                            'AI':   (47.1, 72.4, 26.7, 93.4, 61.9, 44.8, 57.7)  }
```

```python
def plot_by_modality(data, thing, top):
    fig = plt.figure(layout='tight', dpi=300)
    ax = fig.add_subplot(1, 1, 1)

    bars = ax.bar(np.arange(len(participants)) + -0.2, data['Code'], width=0.2)
    bars2 = ax.bar(np.arange(len(participants)), data['GUI'], width=0.2)
    bars3 = ax.bar(np.arange(len(participants)) + 0.2, data['AI'], width=0.2)

    ticks = ax.set_xticks(np.arange(len(participants)), labels=participants)
    ax.set_ylim(top=top)
    text = ax.set_title(f'Plottery {thing} by Modality', fontweight='bold')

    legend = ax.legend(labels=data.keys(), loc=(0.013, 0.92), ncol=3)
    for i, bars in enumerate([bars, bars2, bars3]):
        texts = ax.bar_label(bars, padding=1, color=(0.5, 0.5, 0.5), fontsize='small', rotation='vertical')

    ax.set_ylabel(thing)

    plt.show()

plot_by_modality(interaction_counts, 'Interaction Counts', 18)
plot_by_modality(interaction_percentages, 'Time Percentage', 100)
```
