```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
counts_lists = [
    [4,  1, 3, 6, 10],
    [15, 4, 2, 3, 0],
    [3,  5, 7, 7, 2],
]
```

```python
def plot_counts(counts):
    fig = plt.figure(layout='tight')
    ax = fig.add_subplot(1, 1, 1)

    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.spines['left'].set_visible(False)
    ax.spines['bottom'].set_visible(False)

    ax.set_xticks([])  # Remove x-axis numbers
    ax.set_yticks([])  # Remove y-axis numbers

    left_color  = (1, 0.4, 0.3)
    right_color = (0, 0.5, 1)

    # Interpolating colors
    colors = [(
        left_color[0] * (1 - i / (len(counts) - 1)) + right_color[0] * (i / (len(counts) - 1)),
        left_color[1] * (1 - i / (len(counts) - 1)) + right_color[1] * (i / (len(counts) - 1)),
        left_color[2] * (1 - i / (len(counts) - 1)) + right_color[2] * (i / (len(counts) - 1))
    ) for i in range(len(counts))]

    bars = ax.bar(np.arange(len(counts)), counts, width=1, color=colors, edgecolor=(0.00, 0.00, 0.00), linewidth=7)
    paths = ax.scatter(sum(i*n for i,n in enumerate(counts)) / sum(counts), 0, s=6000, c=(0.00, 0.00, 0.00))

    ax.set_ylim(bottom=-0.15*max(counts))

    plt.show()

for counts in counts_lists:
    plot_counts(counts)
```
