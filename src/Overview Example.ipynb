```python
import numpy as np
import matplotlib.pyplot as plt

from typing import Iterable
```

```python
counts = [4, 1, 3, 6, 10]
```

```python
counts_lists = [
    [4, 1, 3, 6, 10],
    [1, 4, 4, 9, 6],
    [3, 5, 7, 7, 2],
]
```

```python
def plot_counts(counts):
    mean = sum([i*count for i, count in enumerate(counts)]) / sum(counts)
    
    fig = plt.figure(layout='tight')
    ax = fig.add_subplot(1, 1, 1, frameon=False)

    left_color = (1, 0.4, 0.3)
    right_color = (0, 0.5, 1)

    bars_colors = [
        tuple(left_color[j] + (right_color[j] - left_color[j]) * i / (len(counts) - 1)
              for j in range(3))
        for i in range(len(counts))
    ]

    bars = ax.bar(np.arange(len(counts)), counts, width=1, color=bars_colors, edgecolor=(0.00, 0.00, 0.00), linewidth=3)

    ticks = ax.set_xticks([])
    ticks2 = ax.set_yticks([])
    ax.set_ylim(bottom=-0.15*max(counts))

    paths = ax.scatter(mean, 0, s=3200, c=(0.04, 0.04, 0.04))
    paths2 = ax.scatter(0.5, 0.5)

    plt.show()

for i, counts in enumerate(counts_lists):
    plot_counts(counts)
```
